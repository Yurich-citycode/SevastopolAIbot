## Шаг 2. Токен

Вставь токен в поле ниже и нажми Enter. Он попадёт только в память этой сессии:
в блокнот и в git он не записывается.

Если сохранил токен в Colab (значок ключа слева → Secrets → `TG_TOKEN`),
поле можно оставить пустым — возьмётся оттуда.

```python
import os
from getpass import getpass

token = None
try:
    from google.colab import userdata
    token = userdata.get("TG_TOKEN")
except Exception:
    token = None

if not token:
    token = getpass("Токен тестового бота: ")

os.environ["TG_TOKEN"] = (token or "").strip()
print("токен задан:", bool(os.environ["TG_TOKEN"]))
```

## Шаг 5. Открыть бота в Telegram

1. В @BotFather: `/mybots` → твой тестовый бот → **Open** (или просто найди его
   по юзернейму в поиске Telegram).
2. Нажми **Start**.
3. Придёт карточка: Где поесть · Локации · События · Маршруты · Паспорт.
4. Нажимай строки — карточка меняется в том же сообщении, новых сообщений нет.

Если карточка не пришла или вместо неё текст — напиши `/start` ещё раз.

## Если что-то не так

| Симптом | Что делать |
|---|---|
| `Unauthorized` | токен скопирован не целиком, начни с шага 2 |
| пришёл обычный текст вместо карточки | обнови Telegram до последней версии |
| `Terminated` через пару часов | Colab усыпил сессию — запусти шаги 1, 2 и 4 заново |
| бот молчит | проверь, что ячейка шага 4 всё ещё крутится (слева от неё кружок) |
| «карта не отвечает» в разделе «Живая карта» | сервис карты в Colab не запущен, это нормально: остальные разделы работают |

## Важно

Colab — это временная машина: закрыл вкладку или уснула сессия — бот остановился.

запускать на своём компьютере (`python testbotwidgets.py`) или на сервере
