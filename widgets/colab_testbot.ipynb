## Шаг 2. Токен

Ничего вписывать в код не нужно. Запусти ячейку — появится поле ввода,
вставь туда токен и нажми Enter.

Строка токена выглядит так: `8123456789:AAH...` — цифры, двоеточие, дальше
набор букв и цифр. Если вставить его в кавычки внутри кода, бот получит пустую
строку и напишет «нет связи» — токен читается только из поля ввода.

Если сохранил токен в Colab (значок ключа слева → Secrets → имя `TG_TOKEN`,
значение — сам токен), поле можно оставить пустым.

```python
import os
import re
from getpass import getpass

token = ""
try:
    from google.colab import userdata          # если токен сохранён в Secrets
    token = userdata.get("TG_TOKEN") or ""
except Exception:
    token = ""

if not token.strip():
    token = getpass("Вставь токен и нажми Enter: ")

token = (token or "").strip()
os.environ["TG_TOKEN"] = token

looks_right = bool(re.fullmatch(r"\d{6,12}:[A-Za-z0-9_-]{30,}", token))
print("токен задан:", bool(token), "| длина:", len(token), "| формат:", "похож" if looks_right else "НЕ похож")

if token and not looks_right:
    print("Похоже, это не токен. Он выглядит так: 8123456789:AAH...")
elif not token:
    print("Пусто. Запусти ячейку снова и вставь токен в поле ввода, а не в код.")

del token
```
